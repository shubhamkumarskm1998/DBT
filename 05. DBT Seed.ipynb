# 🚀 DBT Seeds & Analysis Folder --- Complete Detailed Guide (Real Production Usage)

------------------------------------------------------------------------

# 🧠 1. What are DBT Seeds?

👉 **DBT Seeds = Static CSV files** that DBT loads into your data
warehouse as tables.

-   Stored in: `seeds/` folder\
-   Format: CSV\
-   Loaded using: `dbt seed`

------------------------------------------------------------------------

## 📌 Key Idea

> Seeds are used for **small, static, reference data**

------------------------------------------------------------------------

## 🧩 Example

### 📁 seeds/country_codes.csv

``` csv
country_code,country_name
IN,India
US,United States
UK,United Kingdom
```

------------------------------------------------------------------------

## ⚙️ Load Seeds

``` bash
dbt seed
```

👉 Creates table in warehouse:

    schema.country_codes

------------------------------------------------------------------------

# 🔥 Why Use DBT Seeds?

------------------------------------------------------------------------

## 1. Reference Data (Most Common Use)

-   Country codes\
-   Currency mappings\
-   Status lookup tables

------------------------------------------------------------------------

## 2. Business Logic Mapping

-   Customer segments\
-   Risk categories

------------------------------------------------------------------------

## 3. Replace Hardcoding

❌ Bad:

``` sql
CASE
  WHEN country = 'IN' THEN 'India'
```

✅ Good:

``` sql
SELECT *
FROM orders o
JOIN {{ ref('country_codes') }} c
ON o.country = c.country_code
```

------------------------------------------------------------------------

## 4. Small Config Tables

-   Feature flags\
-   Threshold values

------------------------------------------------------------------------

# ⚙️ Seed Configuration

------------------------------------------------------------------------

## properties.yml

``` yaml
seeds:
  my_project:
    country_codes:
      +column_types:
        country_code: string
        country_name: string
```

------------------------------------------------------------------------

## dbt_project.yml

``` yaml
seeds:
  my_project:
    +schema: reference_data
```

------------------------------------------------------------------------

# 🔄 How Seeds Work Internally

1.  Reads CSV\
2.  Infers schema\
3.  Creates table\
4.  Inserts data

------------------------------------------------------------------------

# 🧠 Real Production Example (Seeds)

------------------------------------------------------------------------

## 🎯 Problem

Orders table has country_code only → need country_name

------------------------------------------------------------------------

### Step 1: Create Seed

``` csv
country_code,country_name
IN,India
US,United States
```

------------------------------------------------------------------------

### Step 2: Load Seed

``` bash
dbt seed
```

------------------------------------------------------------------------

### Step 3: Use in Model

``` sql
SELECT
    o.order_id,
    c.country_name
FROM {{ ref('orders') }} o
JOIN {{ ref('country_codes') }} c
ON o.country_code = c.country_code
```

------------------------------------------------------------------------

# 🧠 Best Practices for Seeds

-   Keep seeds small\
-   Avoid frequently changing data\
-   Use for lookup only\
-   Version control CSV

------------------------------------------------------------------------

# 🚫 When NOT to Use Seeds

-   Large datasets\
-   Frequently updated data\
-   Streaming data

------------------------------------------------------------------------

# 🧠 2. What is Analysis Folder in DBT?

👉 Used for: - Ad-hoc queries\
- Exploration\
- Debugging

------------------------------------------------------------------------

## 📁 Location

    analysis/

------------------------------------------------------------------------

## 📌 Key Idea

> Analysis = SQL files NOT materialized

------------------------------------------------------------------------

# 🧩 Example

``` sql
SELECT
    DATE(created_at) AS date,
    SUM(amount) AS revenue
FROM {{ ref('orders') }}
GROUP BY 1
```

------------------------------------------------------------------------

# ⚙️ How to Run

``` bash
dbt compile
```

👉 Run compiled SQL manually

------------------------------------------------------------------------

# 🔥 Why Use Analysis

-   Debugging\
-   Exploration\
-   Validation

------------------------------------------------------------------------

# 🧠 Real Use Cases

## Debug Revenue

``` sql
SELECT SUM(amount) FROM {{ ref('orders') }}
```

------------------------------------------------------------------------

## Data Validation

``` sql
SELECT customer_id, COUNT(*)
FROM {{ ref('orders') }}
GROUP BY customer_id
HAVING COUNT(*) > 1000
```

------------------------------------------------------------------------

# ⚙️ Models vs Analysis

  Feature   Models       Analysis
  --------- ------------ ----------
  Output    Table        No table
  Use       Production   Debug

------------------------------------------------------------------------

# 🧠 Combined Flow

Seeds → Models → Analysis → BI

------------------------------------------------------------------------

# 🎯 Summary

## Seeds

-   Static reference data

## Analysis

-   Ad-hoc SQL

👉 Helps avoid hardcoding and improves debugging
