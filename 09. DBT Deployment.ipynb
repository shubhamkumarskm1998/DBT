# 🚀 DBT Deployment using CI/CD + Production Target (YAML-Based Setup)

This is a **production-level guide** on how DBT is deployed using: -
CI/CD pipelines\
- YAML configurations\
- Multiple environments (dev → prod)

------------------------------------------------------------------------

# 🧠 1. What is CI/CD in DBT?

👉 **CI (Continuous Integration)**\
- Validate code changes\
- Run tests before merge

👉 **CD (Continuous Deployment)**\
- Deploy DBT models to production

------------------------------------------------------------------------

## 📌 Goal

Developer → Git → CI → Tests → CD → Production Warehouse

------------------------------------------------------------------------

# 🏗️ 2. Environment-Based Architecture

------------------------------------------------------------------------

## 🔁 Environments

  Environment   Purpose
  ------------- -------------
  dev           Development
  qa            Testing
  prod          Production

------------------------------------------------------------------------

# ⚙️ 3. profiles.yml (Environment Config)

👉 Controls connection to warehouse

------------------------------------------------------------------------

## 📁 \~/.dbt/profiles.yml

``` yaml
my_dbt_project:
  target: dev

  outputs:
    dev:
      type: databricks
      schema: dev_schema
      threads: 4

    prod:
      type: databricks
      schema: prod_schema
      threads: 8
```

------------------------------------------------------------------------

# 🧠 Key Idea

``` bash
dbt run --target prod
```

------------------------------------------------------------------------

# 🧩 4. dbt_project.yml (Deployment Config)

``` yaml
name: my_dbt_project

models:
  my_dbt_project:

    staging:
      +materialized: view
      +schema: staging

    marts:
      +materialized: table
      +schema: analytics
```

------------------------------------------------------------------------

# ⚙️ 5. CI Pipeline (Example YAML)

``` yaml
name: DBT CI Pipeline

on:
  pull_request:
    branches: [main]

jobs:
  dbt-ci:
    runs-on: ubuntu-latest

    steps:
      - uses: actions/checkout@v3
      - uses: actions/setup-python@v4
      - run: pip install dbt-databricks
      - run: dbt deps
      - run: dbt debug
      - run: dbt test
      - run: dbt run -s state:modified
```

------------------------------------------------------------------------

# ⚙️ 6. CD Pipeline (Production Deployment)

``` yaml
name: DBT CD Pipeline

on:
  push:
    branches: [main]

jobs:
  dbt-deploy:
    runs-on: ubuntu-latest

    steps:
      - uses: actions/checkout@v3
      - uses: actions/setup-python@v4
      - run: pip install dbt-databricks
      - run: dbt deps
      - run: dbt run --target prod
      - run: dbt test --target prod
```

------------------------------------------------------------------------

# 🔐 7. Secrets Management

``` yaml
env:
  DBT_TOKEN: ${{ secrets.DBT_TOKEN }}
```

# 🏗️ 9. Production Flow

Developer → PR → CI → Merge → CD → Production

------------------------------------------------------------------------

# ⚡ 10. Best Practices

-   Separate schemas per env\
-   Run tests before deploy\
-   Use secrets\
-   Use state-based runs
