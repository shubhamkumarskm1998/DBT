# 🚀 Jinja in DBT --- Complete Detailed Guide (With Examples)

------------------------------------------------------------------------

# 🧠 1. What is Jinja?

👉 **Jinja** is a **templating engine** (like a mini programming layer)
used inside SQL.

-   Allows dynamic SQL generation\
-   Adds programming features to SQL\
-   Used heavily in DBT

------------------------------------------------------------------------

## 📌 Simple Idea

> Jinja = SQL + Programming Logic

------------------------------------------------------------------------

## 🧩 Example

``` sql
SELECT * FROM {{ ref('orders') }}
```

👉 Jinja (`{{ }}`) gets replaced → actual table name

------------------------------------------------------------------------

# 🚀 2. Why Jinja is Used in DBT?

------------------------------------------------------------------------

## 🔥 1. Dynamic SQL Generation

``` sql
SELECT * FROM {{ ref('stg_orders') }}
```

👉 Avoid hardcoding schema/table names

------------------------------------------------------------------------

## 🔁 2. Reusability

👉 Write logic once → reuse everywhere

------------------------------------------------------------------------

## ⚙️ 3. Conditional Logic

👉 Different behavior for dev vs prod

------------------------------------------------------------------------

## 🔄 4. Loops

👉 Avoid repetitive SQL

------------------------------------------------------------------------

## 🧠 5. Macros

👉 Create reusable SQL functions

------------------------------------------------------------------------

# 🧩 3. Jinja Syntax Basics

------------------------------------------------------------------------

## 📌 Expressions

``` sql
{{ expression }}
```

👉 Used to print values

------------------------------------------------------------------------

## 📌 Statements

``` sql
{% statement %}
```

👉 Used for logic (if, loop, etc.)

------------------------------------------------------------------------

# 🔗 4. ref() (Most Important)

``` sql
SELECT * FROM {{ ref('stg_orders') }}
```

👉 Handles: - Dependency - Table resolution - DAG creation

------------------------------------------------------------------------

# 📥 5. source()

``` sql
SELECT * FROM {{ source('raw', 'orders') }}
```

👉 Refers to raw data

------------------------------------------------------------------------

# ⚙️ 6. config()

``` sql
{{ config(materialized='table') }}
```

👉 Sets model behavior

------------------------------------------------------------------------

# 🔁 7. For Loop (Dynamic Columns)

------------------------------------------------------------------------

## 📌 Example

``` sql
{% for col in ['id', 'amount', 'created_at'] %}
    {{ col }}{% if not loop.last %},{% endif %}
{% endfor %}
```

------------------------------------------------------------------------

## 🧠 Output

``` sql
id, amount, created_at
```

------------------------------------------------------------------------

## 🔥 Real Use Case

👉 Selecting multiple columns dynamically

------------------------------------------------------------------------

# 🔀 8. If / Elif / Else Condition

------------------------------------------------------------------------

## 📌 Example

``` sql
SELECT *
FROM orders

{% if target.name == 'dev' %}
LIMIT 100
{% elif target.name == 'test' %}
LIMIT 1000
{% else %}
-- full data
{% endif %}
```

------------------------------------------------------------------------

## 🧠 Use Case

-   Dev → small data\
-   Prod → full data

------------------------------------------------------------------------

# 🔁 9. Variables

------------------------------------------------------------------------

## 📌 Example

``` sql
SELECT *
FROM orders
WHERE country = '{{ var("country", "IN") }}'
```

------------------------------------------------------------------------

## 🧠 Use Case

-   Dynamic filtering\
-   Environment configs

------------------------------------------------------------------------

# 🧠 10. this (Current Model Reference)

``` sql
SELECT MAX(updated_at)
FROM {{ this }}
```

👉 Refers to: - Current model table

------------------------------------------------------------------------

# 🔥 11. Macros (Most Powerful Feature)

------------------------------------------------------------------------

## 🧠 What is Macro?

👉 A reusable SQL function written in Jinja

------------------------------------------------------------------------

## 📁 Define Macro

``` sql
-- macros/calculate_tax.sql

{% macro calculate_tax(amount) %}
    {{ amount }} * 0.18
{% endmacro %}
```

------------------------------------------------------------------------

## 📌 Use Macro

``` sql
SELECT
    order_id,
    {{ calculate_tax('amount') }} AS tax
FROM orders
```

------------------------------------------------------------------------

## 🧠 Output

``` sql
amount * 0.18
```

------------------------------------------------------------------------

# 🔁 12. Advanced Macro Example

------------------------------------------------------------------------

``` sql
{% macro segment(amount) %}
CASE
    WHEN {{ amount }} > 10000 THEN 'High'
    WHEN {{ amount }} > 5000 THEN 'Medium'
    ELSE 'Low'
END
{% endmacro %}
```

------------------------------------------------------------------------

## Use

``` sql
SELECT
    customer_id,
    {{ segment('total_spent') }} AS segment
FROM table
```

------------------------------------------------------------------------

# 🏗️ 13. Real Production Example (Combined)

------------------------------------------------------------------------

``` sql
{{ config(materialized='incremental') }}

SELECT
    customer_id,

    {% for col in ['amount', 'tax'] %}
        SUM({{ col }}) AS total_{{ col }},
    {% endfor %}

    {{ segment('amount') }} AS segment

FROM {{ ref('stg_orders') }}

{% if is_incremental() %}
WHERE updated_at > (SELECT MAX(updated_at) FROM {{ this }})
{% endif %}

GROUP BY customer_id
```

------------------------------------------------------------------------

# 🧠 14. How Jinja Works Internally

1.  DBT reads SQL file\
2.  Jinja renders templates\
3.  Produces final SQL\
4.  Executes in warehouse
