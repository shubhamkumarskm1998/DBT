# 🚀 DBT CLI Installation with PyCharm + Azure Databricks (Using UV) --- Detailed Notes

------------------------------------------------------------------------

# 🧠 Architecture Overview

DBT follows ELT pattern:

PyCharm (Local) → DBT CLI → Databricks SQL Warehouse → Delta Tables

👉 DBT runs locally\
👉 Databricks executes queries

------------------------------------------------------------------------

# ⚙️ Prerequisites (Detailed)

## 1. Python

-   Version: 3.8+
-   Required because DBT is a Python-based tool

Check:

``` bash
python --version
```

------------------------------------------------------------------------

## 2. PyCharm

-   IDE for development
-   Helps manage virtual env, Git, debugging

------------------------------------------------------------------------

## 3. UV (Why UV instead of pip?)

👉 UV is a faster package manager

Benefits: - Faster dependency resolution - Better performance than pip -
Works with virtual environments

------------------------------------------------------------------------

## 4. Databricks Setup

You need:

-   Workspace URL → used as host
-   SQL Warehouse → execution engine
-   Personal Access Token (PAT) → authentication

------------------------------------------------------------------------

# 🧩 Step 1: Install UV

``` bash
pip install uv
```

Verify:

``` bash
uv --version
```

------------------------------------------------------------------------

# 🧩 Step 2: Create Virtual Environment

👉 Why? - Isolates dependencies - Avoids conflicts

``` bash
uv venv
```

Activate:

``` bash
# Windows
.venv\Scripts\activate

# Mac/Linux
source .venv/bin/activate
```

------------------------------------------------------------------------

# 🧩 Step 3: Install DBT (Databricks Adapter)

``` bash
uv pip install dbt-databricks
```

👉 Why adapter? - DBT needs adapter to communicate with warehouse

Verify:

``` bash
dbt --version
```

------------------------------------------------------------------------

# 🧩 Step 4: Initialize DBT Project

``` bash
dbt init my_dbt_project
```

👉 What happens internally: - Creates project structure - Generates
config files - Links to profiles.yml

------------------------------------------------------------------------

# 📁 Folder Structure (Detailed Explanation)

    models/        → Core transformations (SQL files)
    macros/        → Reusable SQL logic (functions)
    tests/         → Custom tests
    snapshots/     → Historical tracking (SCD Type 2)
    seeds/         → Static CSV → tables
    analyses/      → Ad-hoc queries
    target/        → Compiled SQL + DAG metadata
    logs/          → Execution logs
    dbt_project.yml→ Project config

------------------------------------------------------------------------

## 🔥 Important Folders Explained

### models/

-   Main transformation logic
-   Organized in layers:
    -   staging
    -   intermediate
    -   marts

------------------------------------------------------------------------

### macros/

-   Jinja-based reusable logic
-   Avoid repetition

------------------------------------------------------------------------

### target/

-   Auto-generated by DBT
-   Contains:
    -   compiled SQL
    -   manifest.json (DAG)
    -   run results

------------------------------------------------------------------------

# 🧩 Step 5: Configure profiles.yml (VERY IMPORTANT)

Location:

    ~/.dbt/profiles.yml

------------------------------------------------------------------------

## 🔧 Databricks Configuration

``` yaml
my_dbt_project:
  target: dev
  outputs:
    dev:
      type: databricks
      method: http
      host: adb-xxxx.azuredatabricks.net
      http_path: /sql/1.0/warehouses/xxxx
      token: dapiXXXX
      catalog: hive_metastore
      schema: default
      threads: 4
```

------------------------------------------------------------------------

## 🔑 Explanation of Fields

-   host → Databricks workspace URL
-   http_path → SQL warehouse endpoint
-   token → authentication
-   catalog/schema → where tables are created
-   threads → parallel execution

------------------------------------------------------------------------

# 🧩 Step 6: Test Connection

``` bash
dbt debug
```

👉 What it checks: - Profile config - Warehouse connection - Credentials

------------------------------------------------------------------------

# 🧩 Step 7: Create First Model

``` sql
-- models/example.sql
SELECT 1 AS id
```

👉 Model = SQL file → becomes table/view

------------------------------------------------------------------------

# 🧩 Step 8: Run DBT

``` bash
dbt run
```

👉 Internally: 1. Parse project 2. Build DAG 3. Compile SQL 4. Send to
Databricks 5. Warehouse executes

------------------------------------------------------------------------

# 🧩 Step 9: Add Dependency Using ref()

``` sql
SELECT * FROM {{ ref('stg_orders') }}
```

👉 Creates DAG: stg_orders → downstream model

------------------------------------------------------------------------

# 🧩 Step 10: Run Tests

``` bash
dbt test
```

👉 DBT generates SQL to validate: - nulls - duplicates

------------------------------------------------------------------------

# 🧩 Step 11: Generate Documentation

``` bash
dbt docs generate
dbt docs serve
```

👉 Shows: - DAG lineage - model relationships

------------------------------------------------------------------------

# ⚡ Databricks-Specific Notes

## 1. Uses SQL Warehouse

-   Not Spark cluster directly
-   DBT sends SQL queries

------------------------------------------------------------------------

## 2. Delta Tables

-   Stored in Delta format
-   Supports ACID + performance

------------------------------------------------------------------------

## 3. Incremental Models (Important)

``` sql
{{ config(materialized='incremental') }}
```

👉 Only processes new data → improves performance

------------------------------------------------------------------------

# 🔥 Common Errors & Fixes

## ❌ dbt not found

👉 Activate virtual environment

------------------------------------------------------------------------

## ❌ Connection error

-   Check token expiry
-   Check http_path

------------------------------------------------------------------------

## ❌ Permission issue

-   Ensure schema access in Databricks

------------------------------------------------------------------------

# 🧠 Best Practices

-   Use separate dev/prod schemas
-   Use Git for version control
-   Use incremental models for large data
-   Keep secrets secure (don't hardcode tokens)

------------------------------------------------------------------------

# 🚀 Production Architecture

Airflow → DBT CLI → Databricks → BI Tools
